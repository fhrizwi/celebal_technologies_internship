Missing Values

Find missing values

```python
df.isnull().sum()
```

If numerical column

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

If text column

```python
df.fillna("Unknown", inplace=True)
```

Drop rows

```python
df.dropna(inplace=True)
```

Filter Rows

```python
df[df["Quantity"] > 5]
```

Select Columns

```python
df[["Product Name","Sales"]]
```

Remove Duplicates

```python
df.drop_duplicates(inplace=True)
```

Create Derived Column

```python
df["total_amount"] = df["Sales"] * df["Quantity"]
```

Save Cleaned CSV

```python
df.to_csv("../output/cleaned_store.csv", index=False)
```
