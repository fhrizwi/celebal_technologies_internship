```python
from pyspark.sql.types import *
from pyspark.sql.functions import *

call_schema = StructType([
    StructField("call_id", StringType()),
    StructField("customer_id", IntegerType()),
    StructField("call_timestamp", TimestampType()),
    StructField("duration_seconds", IntegerType()),
    StructField("sentiment", StringType()),
    StructField(
        "conversation",
        StructType([
            StructField("agent", StringType()),
            StructField("customer_text", StringType())
        ])
    )
])
```

```python
stream_df = (
    spark.readStream
        .schema(call_schema)
        .option("multiLine", True)
        .json("/Volumes/workspace/default/data/call_data")
)
```

```python
watermark_df = (
    stream_df
        .withWatermark("call_timestamp", "10 minutes")
        .dropDuplicates(["call_id"])
)
```

```python
print(watermark_df.isStreaming)
```
