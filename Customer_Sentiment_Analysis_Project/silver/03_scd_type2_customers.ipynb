```python
silver_df = spark.table("workspace.default.silver_customers")
```

```python
from pyspark.sql.functions import lit, col

scd_df = (
    silver_df
        .withColumn("effective_from", col("update_ts"))
        .withColumn("effective_to", lit(None).cast("timestamp"))
        .withColumn("is_current", lit(True))
)
```

```python
(
    scd_df.write
        .format("delta")
        .mode("overwrite")
        .saveAsTable("workspace.default.dim_customers_scd2")
)
```

```python
display(
    spark.sql("""
    SELECT *
    FROM workspace.default.dim_customers_scd2
    LIMIT 20
    """)
)
```
