```python
import random
import uuid
import json
from datetime import datetime

customer_ids = list(range(1,101))

sentiments = [
    "Positive",
    "Negative",
    "Neutral"
]

sample_text = {
    "Positive":[
        "Issue resolved successfully",
        "Very happy with service",
        "Excellent support"
    ],
    "Negative":[
        "Very disappointed",
        "Problem not resolved",
        "Bad experience"
    ],
    "Neutral":[
        "Need more information",
        "Call transferred",
        "Waiting for update"
    ]
}
```

```python
def generate_call():

    sentiment = random.choice(sentiments)

    return {

        "call_id": str(uuid.uuid4()),

        "customer_id": random.choice(customer_ids),

        "call_timestamp": datetime.now().isoformat(),

        "duration_seconds": random.randint(30,1200),

        "sentiment": sentiment,

        "conversation":{

            "agent":"Agent_"+str(random.randint(1,20)),

            "customer_text":random.choice(sample_text[sentiment])

        }

    }
```

```python
record = generate_call()

print(record)
```

```python
import os
import time

output_path = "/Volumes/workspace/default/data/call_data"
```

```python
def save_call_json():

    record = generate_call()

    file_name = f"{record['call_id']}.json"

    full_path = os.path.join(output_path, file_name)

    with open(full_path, "w") as f:
        json.dump(record, f, indent=4)

    print("Created:", file_name)
```

```python
save_call_json()
```

```python
for i in range(10):

    save_call_json()

    time.sleep(2)
```

```python
while True:

    save_call_json()

    time.sleep(5)
```
